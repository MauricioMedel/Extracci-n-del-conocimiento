## algortimo KNN

```python
#importamos lass variables 


import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
```

```python
iris = load_iris(as_frame=True)
```

```python
print(iris.DESCR)
```

```python
dfIris = iris.frame
dfIris.head(2)
```

```python
dfIris.tail(2)
```

```python
#Cambiamos la columna de Target por Especie
dfIris.rename(columns={"target" : "species"}, inplace=True)
```

```python
#Remplazamos los valores numericos por las especies
dfIris["species"] = dfIris["species"].map(
    {
        0: "setosa",
        1: "versicolor",
        2: "virginica"
    }
)
```

```python
dfIris.head(5)
```

## variables preditoras y variable objetivo 

x variables prectoras que son aquellas que me permiten hacer la prediccion 
y es el objetivo que quiero predecir 

```python
X = dfIris.drop(columns=["species"])
```

```python
y = dfIris["species"]
```

```python
print("----Variables predictoras----")
print(X.columns)
print("----Variable Objetivo -----")
print(y.name)
```

## División del conjunto de datos 

Training set: Datos para entrenamiento 80%

Testing Set: Datos no vistos por el modelo para pruebas 20%

```python
# Solución: Agregamos y_test al final
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)
```

```python
print("X_train: ", X_train.shape)
print("X_test: ", X_test.shape)
print("-------------------------")
print("y_train: ", y_train.shape)
print("y_test: ", y_test.shape)
```

## Escalado de las variables 

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("-------------Datos sin escalar-------------")
print(X_train)
print("-------------Datos escalados-------------")
print(X_train_scaled)
```

## entrenamiento del modelo 

```python


#Creamos una instancia del algoritmo
modelo = KNeighborsClassifier(
    n_neighbors = 3
)
```

```python
#Entrenamos con los datos de entrenamiento

modelo.fit(
    X_train_scaled,
    y_train
)
```

## prediccion con los datos de prueba 



```python
prediccion = modelo.predict(X_test_scaled)


pd.DataFrame({
    "Especie real" : y_test.values,
    "Especie predicha" : prediccion
}).head(20)
```

## Evaluación del modelo 

*las metricas a usar son: accuracy, precision, recall, f2-score*

```python
accuracy = accuracy_score (y_test, prediccion)
```

```python
print(f"Accuracy / Exactitud del modelo es: {accuracy:.4f}")
```

```python
matrizConfusion = confusion_matrix(y_test, prediccion)
matrizConfusion_df = pd.DataFrame(
    matrizConfusion,
    index= modelo.classes_,
    columns = modelo.classes_
)
```

```python
matrizConfusion_df
```

```python
print (classification_report(y_test, prediccion))
```
