```python
import pandas as pd
import numpy as np
import random
```

```python
from faker import Faker
```

```python
pip install Faker
```

```python
faker = Faker("es_MX")
```

```python
datos = []
for i in range (100):
    datos.append (
        {
            "id_usuario" : i + 1,
            "nombre" : faker.name(),
            "correo" : faker.email(),
            "edad" : random.randint(18,65),
            "seguidores" : random.randint(50,5000),
            "plataforma" : random.choice(["tiktok", "instagram","YouTube" ])
        }
)
```

```python
dfInfluencer = pd.DataFrame(datos)
```
