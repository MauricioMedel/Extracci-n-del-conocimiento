```python
import pandas as pd
```

```python
# Conjunto de datos base para la generación del Big Data
datos = {
    "usuario": ["ana", "pepe", "luis", "rosita", "carolina", "paco"],
    "edad": [14, 22, 35, 40, 62, 70],
    "ciudad": ["cuitlahuac", "yanga", "atoyac", "paraje", "fortin", "cuichapa"],
    "compras_realizadas": [5, 10, 1, 2, 15, 3],
    "activo": [True, True, True, False, True, False],
    "fecha_registro": ["2026-01-10", "2026-05-21", "2026-02-10", "2026-03-11", "2026-04-10", "2026-05-01"]
}
```

```python
#Creamos el dataframe
dfBigData = pd.DataFrame(datos)
```

```python
dfBigData
```

```python
#cantidada de registros 
cantidad = 1000000

```

```python
import random
```

```python
#crear el dataframe a 1000000 de registros
```

```python
# Crear el dataFrame de 1000000 de registros
dfMillon = pd.DataFrame({
    "nombre": [random.choice(dfBigData["usuario"]) for _ in range(cantidad)],
    "edad": [random.randint(18, 80) for _ in range(cantidad)],
    "ciudad": [random.choice(dfBigData["ciudad"]) for _ in range(cantidad)],
    "compras_realizadas": [random.randint(1, 10) for _ in range(cantidad)],
    "activo": [random.choice(dfBigData["activo"]) for _ in range(cantidad)],
    "fecha_registro": [random.choice(dfBigData["fecha_registro"]) for _ in range(cantidad)]
})

```
